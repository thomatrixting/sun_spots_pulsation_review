# 03A — Sunspot pulsation analysis: DS01

Example notebook that applies the full pipeline from `src/sunspot_analysis.py` to the **DS01** region.

**Workflow**
1. Verify cadence from the `.sav` timing file
2. Load cubes + build umbra / penumbra masks (adjust thresholds if needed)
3. Calibration frame — visual check of the masks
   - `plot_histogram(data)` — full-cube value histogram; use `cube=` to switch between magnetogram / continuum / dopplergram
   - `plot_magnetogram_masks(data)` — magnetogram with `sun_spot` and `hot_spot` overlays
4. Compute per-frame metrics
5. Save metrics CSV
6. Time-series plots + area vs time
7. FFT plots (separate + combined with zoom)
8. Save animation

```python
#%load_ext autoreload   
#%autoreload 2

import sys
import pathlib

project_root = pathlib.Path.cwd()
if not (project_root / 'src').exists():
    project_root = project_root.parent
sys.path.insert(0, str(project_root.resolve()))

import numpy as np
import matplotlib.pyplot as plt

from src.sunspot_analysis import (
    verify_cadence,
    load_and_mask,
    plot_calibration_frame,
    plot_histogram,
    plot_magnetogram_masks,
    plot_area,
    compute_metrics,
    save_metrics_csv,
    plot_time_series,
    plot_ffts_separate,
    plot_ffts_combined,
    save_animation,
)
```

# DS00

```python
# ── paths ─────────────────────────────────────────────────────────────────────
DS0N_NAME      = 'DS00'
raw_dir_0N      = pathlib.Path('../data/raw/sebastian_sun_spots') / DS0N_NAME
processed_dir_0N = pathlib.Path('../data/processed/sebastian_sun_spots') / DS0N_NAME
plots_dir_0N    = processed_dir_0N / 'plots'

print('Raw data dir :', raw_dir_0N)
print('Processed dir:', processed_dir_0N)
```

## Step 0 — Verify cadence from the `.sav` timing file

```python
cadence_s, cadence_list = verify_cadence(raw_dir_0N)
arr, counts = np.unique(cadence_list, return_counts=True)
dict(zip(arr, counts))
```

## Step 1 — Load cubes and build region masks

Adjust `umbra_thresh` and `penumbra_thresh` after looking at the calibration frame below.
Run **Steps 1–2** together until the masks look right, then proceed.

```python
mag_filter_0N = lambda x : x>500

data = load_and_mask(
    ds_dir          = raw_dir_0N,
    umbra_thresh    = 30_000,   # continuum DN — pixels below this → umbra
    penumbra_thresh = 50_000,   # continuum DN — upper bound for penumbra
    cadence_s       = 720,
    mag_filter=mag_filter_0N,
    filter_mask=False
)

print(f"Cube shape (n_t, ny, nx): {data['cube_cont'].shape}")
print(f"Time span: {data['time_h'][-1]:.1f} h  ({data['n_t']} frames @ {data['cadence_s']:.0f} s)")
```

## Step 2 — Calibration frame

Change `frame_idx` to inspect any frame.  
Green overlay = umbra, purple = penumbra.

```python
# ── 2b: value histogram — inspect cube distribution ─────────────────────────
# Plots the full-cube histogram (all pixels × all frames).
# Change cube= to 'continuum' or 'dopplergram' to inspect those cubes.
# Use the printed percentile table to choose a mag_filter threshold.
plot_histogram(
    data,
    cube      = 'magnetogram',
    save      = False,
    plots_dir = plots_dir_0N,
)
```

```python
# ── 2c: magnetogram with region overlays ─────────────────────────────────────
# Purple = continuum-based sun_spot (umbra + penumbra).
# Orange = hot_spot (mag_filter region). Only shown if load_and_mask was called
# with a mag_filter; otherwise only the purple sun_spot overlay appears.
plot_magnetogram_masks(
    data,
    frame_idx = 500,    # None → middle frame
    save      = False,
    plots_dir = plots_dir_0N,
)
```

```python
plot_calibration_frame(
    data,
    frame_idx = 0,       # use None for the middle frame, or any integer
    save      = False,
    plots_dir = plots_dir_0N,
)
```

```python
# Intermediate frame check — useful when the sunspot evolves significantly
plot_calibration_frame(
    data,
    frame_idx = 500,
    save      = True,
    plots_dir = plots_dir_0N,
)
```

## Step 3 — Compute per-frame metrics

```python
metrics = compute_metrics(data)

n_pix = data['umbra'].sum(axis=(1, 2))
print(f"Umbra pixels/frame  →  mean={n_pix.mean():.0f}, min={n_pix.min()}, max={n_pix.max()}")
print(f"Mean B   umbra    : {np.nanmean(metrics['mean_mag_umb']):.1f} G")
print(f"Mean B   penumbra : {np.nanmean(metrics['mean_mag_pen']):.1f} G")
print(f"Mean v   umbra    : {np.nanmean(metrics['mean_dop_umb']):.1f} m/s")
print(f"Mean v   penumbra : {np.nanmean(metrics['mean_dop_pen']):.1f} m/s")
```

## Step 4 — Save metrics CSV

```python
save_metrics_csv(data, metrics, processed_dir_0N)
```

## Step 5 — Time-series plots

```python
# ── area vs time ─────────────────────────────────────────────────────────────
# Pixel count of each region per frame. If hot_spot is defined it appears in red.
plot_area(data, metrics, save=True, plots_dir=plots_dir_0N)
```

```python
plot_time_series(data, metrics, save=True, plots_dir=plots_dir_0N)
```

```python
plot_time_series(data, metrics, normalized=True, save=True, plots_dir=plots_dir_0N)
```

## Step 6 — FFT analysis

### 6a — Separate subplots (one per region)

```python
plot_ffts_separate(
    data, metrics,
    annotate_peaks = True,
    print_table    = True,
    save           = False,
    plots_dir      = plots_dir_0N,
)
```

### 6b — Combined plot with x-axis zoom

Adjust `xlim` to focus on the frequency range of interest (in mHz).  
The peak-table CSV (`fft_peaks.csv`) is saved alongside the figure.

```python
plot_ffts_combined(
    data, metrics,
    xlim           = (0, 0.1),   # mHz — change or set to None for full range
    annotate_peaks = True,
    print_table    = True,
    save           = True,
    plots_dir      = plots_dir_0N,
)
```

## Step 7 — Save animation

```python
anim_path = processed_dir_0N / 'anim.html'

save_animation(
    data, metrics,
    save_path = anim_path,
    step      = 50,   # include every 50th frame
    fps       = 5, 
    mag_symmetric_cbar=True
)
```

```python
# Open the saved animation in the browser
from IPython.display import FileLink
FileLink(str(anim_path))
```

```python
from scipy.signal import argrelextrema

fig, ax1 = plt.subplots(figsize=(12,8))

plt.title('Mean values for the Umbra')

mag = np.asarray(metrics['mean_mag_umb'])
t = np.asarray(data['time_h'])

ax1.plot(t, mag, label='mean magnetic hotspot', color='r')
ax1.set_ylabel('Mean B(G)')
ax1.tick_params(axis='y', labelcolor='r')

# Local maxima and minima
max_idx = argrelextrema(mag, np.greater,order=25)[0]
min_idx = argrelextrema(mag, np.less,order=25)[0]

for i in max_idx:
    ax1.axvline(t[i], color='g', linestyle='--', alpha=0.7,
                label='local max' if i == max_idx[0] else None)
for i in min_idx:
    ax1.axvline(t[i], color='m', linestyle=':', alpha=0.7,
                label='local min' if i == min_idx[0] else None)

ax2 = ax1.twinx()
ax2.plot(t, metrics['mean_dop_umb'], label='mean dopler', color='b')
ax2.set_ylabel('Mean Velocity (m/s)')
ax2.tick_params(axis='y', labelcolor='b')

ax1.set_xlabel('time(h)')
fig.legend(loc='center left', bbox_to_anchor=(1.05, 0.5))
fig.tight_layout()
plt.show()
```
